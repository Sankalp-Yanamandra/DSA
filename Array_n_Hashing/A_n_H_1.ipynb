#### Find the Largest element in an array
#### Problem Statement: Given an array, we have to find the largest element in the array.

```python
'''method 1 : 1st pass of bubble sort : bubbled at end
    modifies original array
'''
def find_largest(arr):
    for i in range(0,len(arr)):
        for j in range(0,len(arr)-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
        
        if i==0:
            return arr[-1]

arr = list((map(int,input().split())))
print(f'Largest in {arr} : {find_largest(arr)}')

'''
method 2 : doesn't modify original
'''

def find_largest(arr):
    highest = float('-inf')

    for i,num in enumerate(arr):
        if num > highest:
            highest = num
    return highest

arr = list((map(int,input().split())))
print(f'Largest in {arr} : {find_largest(arr)}')
```

Find Second Smallest and Second Largest Element in an array

Problem Statement: Given an array,
find the second smallest and second largest element in the array. Print '-1' in the event that either of them doesn't exist.


```python
'''
method 1 : 
'''

def second_largest_smallest(arr):
    n = len(arr)

    if n in {0,1}:
        return (-1, -1)
    
    largest = max(arr)
    smallest = min(arr)

    # 2nd largest
    second_largest = float('-inf')

    # 2nd smallest
    second_smallest = float('inf')

    for ele in arr:
        if largest > ele > second_largest:
            second_largest = ele

        if second_smallest > ele > smallest:
            second_smallest = ele

    if second_largest == float('-inf') or second_smallest == float('inf'):
        # all same ele
        return (-1,-1)

    return second_smallest, second_largest


arr = list((map(int,input().split())))
small, large = second_largest_smallest(arr)
print(f'2nd Smallest in {arr} : {small} ||| 2nd Largest in {arr} : {large}')

'''
method 2 : best
'''
def second_max_min(arr):
    n = len(arr)

    if n < 2:
        return (-1,-1)
    

    largest = float('-inf')
    sec_largest = float('-inf')

    smallest = float('inf')
    sec_smallest = float('inf')

    for ele in arr:
        # find 2nd largest
        if ele > largest:
            sec_largest = largest
            largest = ele
        elif ele > sec_largest and ele != largest:
            sec_largest = ele

        # find 2nd smallest
        if ele < smallest:
            sec_smallest = smallest
            smallest = ele
        elif ele < sec_smallest and ele != smallest:
            sec_smallest = ele

    if sec_largest == float('-inf') or sec_smallest == float('inf'):
        return (-1, -1)

    return sec_smallest, sec_largest

arr = list((map(int,input().split())))
small, large = second_max_min(arr)
print(f'2nd Smallest in {arr} : {small} ||| 2nd Largest in {arr} : {large}')
```

Check if an Array is Sorted

Problem Statement: Given an array of size n, write a program to check if the given array is sorted in (ascending / Increasing / Non-decreasing) order or not. If the
array is sorted then return True, Else return False.

```python
'''
my method : check 1st pass of bubble sort
'''

def is_sorted(arr):
    n = len(arr)

    if n < 2:
        return True

    for i in range(0,n):
        swapped = False

        for j in range(0,n-1):
            if arr[j] > arr[j+1]:
                swapped = True
                return False # swaps, not sorted

        # end of 1st pass
        if swapped == False:
            # sorted, no swaps
            return True

arr = list(map(int,input().split()))
print(f'{arr} is sorted : {is_sorted(arr)}')
```
