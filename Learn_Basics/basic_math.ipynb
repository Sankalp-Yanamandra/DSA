```python
'''Problem Statement: Given an integer N, return the number of digits in N.'''

'''brute force : O(length of number) = O(log10N)'''
n = int(input('num = '))

def length(n):
    if n==0:
        return 1
    else:
        n = abs(n)# for negative integers
        count = 0
        while n != 0:
            count += 1
            n//=10
        return count

print(f'len of {n} = {length(n)}')

'''optimal : O(1) : logarithm formula'''

import math
def length(n):

    return int(math.log(abs(n),10) + 1)

print(f'len of {n} = {length(n)}')

```

```python
'''Reverse Digits of A Number

Problem Statement: Given an integer N return the reverse of the given number.

Note: If a number has trailing zeros, then its reverse will not include them. 
For e.g , reverse of 10400 will be 401 instead of 00401.'''


# time : O(length of number) = O(log10N)
number = int(input('num = '))

import math
def reverseNum(n):
    length = math.floor(math.log(n,10) + 1)

    #edge case : 0-9
    if length == 1:
        return n

    pos = -1
    rev = 0

    while n!=0:
        digit = n%10
        pos += 1
        place_val = 10 ** (length - pos - 1)
        rev += (digit * place_val)

        n = n //10

    return rev

print(f'reverse of {number} : {reverseNum(number)}')
```

```python
'''GCD of a,b'''

a,b = map(int, input().split(','))

def GCD(a,b):
    if b==0:
        return a
    else:
        return GCD(b, a%b)

print(f'GCD({a},{b}) = {GCD(a,b)}')

```

```python
'''
Check if a number is prime or not

Problem Statement: Given an integer N, check whether it is prime or not. 
A prime number is a number that is only divisible by 1 and itself 
and the total number of divisors is 2..
'''

'''Brute force check all integers upto n : O(n)'''
num = int(input())

def isPrime(n):
    # edge case
    if n < 2:
        return False
    elif n == 2:
        return True
    else:

        for integer in range(2, n):
            if n % integer == 0:
                return False
        return True

if isPrime(num):
    print(f'{num} : prime')
else:
    print(f'{num} : composite')


'''Optimized : check all integers upto sqroot(n) : O(sqroot(n))'''
num = int(input())
import math
def isPrime(n):
    # edge case
    if n < 2:
        return False
    elif n == 2:
        return True
    else:
        limit = math.floor(math.sqrt(n)) + 1
        for integer in range(2, limit):
            if n % integer == 0:
                return False
        return True

if isPrime(num):
    print(f'{num} : prime')
else:
    print(f'{num} : composite')

```

```python
'''
Print all Divisors/factors of a given Number

Problem Statement: Given an integer N, return all divisors of N.
A divisor of an integer N is a positive integer that divides N without leaving a remainder.
In other words, if N is divisible by another integer without any remainder, 
then that integer is considered a divisor of N.
'''

'''brute force : O(n)'''
num = int(input())

def factors(n):
    divisors = []

    for integer in range(1, n+1):
        if n%integer == 0:
            divisors.append(integer)

    return divisors

print(f'divisors of {num} : {factors(num)}')


'''optimized : O(sqroot(n))'''
num = int(input())
import math
def factors(n):
    divisors = set()

    limit = math.floor(math.sqrt(n)) + 1

    for integer in range(1, limit):
        if n%integer == 0:
            divisors.add(integer)
            divisors.add(n//integer)

    return divisors

print(f'divisors of {num} : {factors(num)}')

```

```python
'''
Check if a number is Armstrong Number or not

Problem Statement:Given an integer N, 
return true it is an Armstrong number otherwise return false.

An Amrstrong number is a number that is equal to 
the sum of its own digits each raised to the power of the number of digits.
'''

n = int(input())
import math
def isArmstrong(n):
    if n < 0:
        return False
    if n == 0:
        return True

    length = math.floor(math.log(n, 10)) + 1

    temp = n
    digits_sum = 0
    while temp != 0:
        digits_sum += (temp%10)**length
        temp //= 10
    return digits_sum == n

print(f'{n} is Armstrong : {isArmstrong(n)}') 

```
