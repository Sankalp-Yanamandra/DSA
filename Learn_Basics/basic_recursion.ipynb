```python
'''
Print Name N times using Recursion

Problem Description: Given an integer N, write a program to print your name N times.
'''

def recursiveNameCall(i,name,N):
    # base case
    if i>N:
        return
    else:
        #recursive case
        print(name, end=" ")
        # recursive call
        recursiveNameCall(i+1, name,N)


name = input()
N = int(input())

recursiveNameCall(1, name, N)
```

```python
'''
Print 1 to N using Recursion

Problem Description: Given an integer N, write a program to print numbers from 1 to N.
'''


# forward recursion : print() -> recursive call
def oneToN(i,N):
    if i==N:
        print(i)
        return

    print(i, end=",")

    oneToN(i+1, N)

    
N = int(input())

oneToN(1, N) # output : 1,2,3,4,5

# Backtracking : recursive call -> print()
def oneToN(i,N):
    if i>N:
        return

    oneToN(i+1, N)

    print(i, end=",")

    
N = int(input())

oneToN(1, N) # output : 5,4,3,2,1,

```

```python
'''
Print N to 1 using Recursion

Problem Description: Given an integer N, write a program to print numbers from N to 1.
'''

# forward recursion : print() -> recursive call
def oneToN(i):
    if i==0:
        return

    print(i, end=",")

    oneToN(i-1)

    
N = int(input())

oneToN(N)

print()

# Backtracking : recursive call -> print()
def oneToN(i):
    if i==0:
        return

    oneToN(i-1)

    print(i, end=",")

    
N = int(input())

oneToN(N)

```

```python
'''
Sum of first N Natural Numbers

Problem Statement: Given a number ‘N’, find out the sum of the first N natural numbers .
'''

# my beginner friendly aprroach
def firstNsum(i, running_sum, N):
    if i>N:
        return running_sum

    running_sum += i
    return firstNsum(i+1,running_sum,N)

N = int(input())
ans = firstNsum(0,0,N)
print(f'sum of first {N} natural nos = {ans}')

# must better
def firstNsum(i):
    if i==1:
        return 1

    return i + firstNsum(i-1)

N = int(input())
ans = firstNsum(N)
print(f'sum of first {N} natural nos = {ans}')

```

```python
'''
Factorial of a Number : Iterative and Recursive

Problem Statement: Given a number X,  print its factorial.

To obtain the factorial of a number, it has to be multiplied by all the whole numbers preceding it.
More precisely X! = X*(X-1)*(X-2) … 1.

Note: X  is always a positive number. 
'''

def fact(N):
    if N == 0 or N == 1:
        return 1

    return N * fact(N-1)

N = int(input())

print(f'{N}! = {fact(N)}')
```

```python
'''
Reverse a given Array

Problem Statement: You are given an array. The task is to reverse the array and print it.
'''

# brute force : my way

def reverseArray(old, ptr, new):
    if ptr < - len(old):
        print(new)
        return
    else:
        new.append(old[ptr])
        reverseArray(old, ptr - 1, new)

old = list(map(int, input().split()))
reverseArray(old, -1, [])

# better : 2 ptr O(n/2)
def reverseArray(old):
    left, right = 0, len(old) - 1

    def recursive(left, right):
        if left > right:
            return
        else:
            old[left], old[right] = old[right], old[left]
            return recursive(left+1, right-1)

    recursive(left, right)
    return old

old = list((map(int,input().split())))
print(f'before : {old}')
reverseArray(old)
print(f'after : {old}')
```

```python
'''
Print Fibonacci Series up to Nth term
Problem Statement: Given an integer N. Print the Fibonacci series up to the Nth term.
'''

# for practice : Nth term
def NthTerm(N):
    if N==0 or N==1:
        return N
    else:
        return NthTerm(N-1) + NthTerm(N-2)

N = int(input()) 
print(f'{N}th term of fibonacci : {NthTerm(N)}')

# my way
def Nterms(first, second, N):
    if N == 1:
        return
    else:
        term = first + second
        print(term, end=",")
        return Nterms(second, term, N-1)

N=int(input())
first,second = 0, 1
print(first, second, sep=",",end=",")
Nterms(first,second,N)
```
