## Iterative Simple/Brute force bubble sort (ascending)

```python
def bruteBubbleSortAsc(array):
    n = len(array)
    # outer loop : n-1 passes
    for iteration in range(0, n-1):
        # inner loop : comparison of n-1 ele and swap if req
        for comparison in range(0, n-1):
            if array[comparison] > array[comparison + 1]:
                # swap
                array[comparison], array[comparison+1] = array[comparison+1], array[comparison]

array = list(map(int,input().split()))
print(f'Before : {array}')
bruteBubbleSortAsc(array)
print(f'after : {array}')
```

## Iterative Optimized bubble sort asc -> best : O(n)

```python
def OptimizedBubbleSortAsc(array):
    n= len(array)
    # outer loop : n-1 passes
    for passes in range(0, n-1):
        # flag to check if no swap
        no_swapping = True
        # inner loop : comparison of (n-1)-i ele i.e. after each pass i ele sorted at end, swap if req
        for comparison in range(0, (n-1)-passes):
            if array[comparison] > array[comparison+1]:
                no_swapping = False
                array[comparison], array[comparison+1] = array[comparison+1],array[comparison]
        if no_swapping == True:
            break # array sorted

array = list(map(int,input().split()))
print(f'before : {array}')
OptimizedBubbleSortAsc(array)
print(f'after : {array}')
```

## Recursive brute force Bubble Sort Ascending

```python
def comparison(arr,j,n):
    # inner loop
    if j>n-2:
        return

    if arr[j] > arr[j+1]:
        arr[j], arr[j+1] = arr[j+1], arr[j]
    comparison(arr,j+1,n)

def recursive_brute_bubbble_sort_asc(arr,i,n):
    # outer loop
    if i>n-1:
        return
    j = 0
    comparison(arr, j, n)
    recursive_brute_bubbble_sort_asc(arr, i+1, n)

arr = [13,46,24,52,20,9]
print(f'before : {arr}')
recursive_brute_bubbble_sort_asc(arr,0,len(arr))
print(f'after : {arr}')
```

## recursive optimized bubble sort

```python
def comparison(arr, j, limit, swapped):
    if j >= limit:
        return swapped

    if arr[j] > arr[j + 1]:
        arr[j], arr[j + 1] = arr[j + 1], arr[j]
        swapped = True

    return comparison(arr, j + 1, limit, swapped)


def recursive_optimized_bubble_sort_asc(arr, i, n):
    if i >= n - 1:
        return

    swapped = comparison(arr, 0, n - 1 - i, False)

    if not swapped:
        return

    recursive_optimized_bubble_sort_asc(arr, i + 1, n)

arr = [13,46,24,52,20,9]
print(f'before : {arr}')
recursive_optimized_bubble_sort_asc(arr,0,len(arr))
print(f'after : {arr}')
```
