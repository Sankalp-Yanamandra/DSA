## Insertion sort Ascending

```python
def insertion_sort_ascending(arr):
    n = len(arr)
    # outer loop : unsorted sublist (doesnt include 1st ele)
    for i in range(1,n):
        # inner loop : compare with sorted sublist to find the correct posn
        temp = arr[i]
        j = i - 1
        while j>=0 and arr[j] > temp:
            arr[j+1] = arr[j]
            j -= 1

        arr[j+1] = temp

arr = list(map(int,input().split())) 
print(f'before : {arr}')
insertion_sort_ascending(arr)
print(f'after : {arr}')
```

### Recursive insertion sort ascending

```python
def find_insert_pos(arr,temp,j,n):
    if not (j>=0 and arr[j] > temp):
        return j
        # since j: int (immutable) : pass by obj value, so need to return updated j

    # shift ele Right by 1
    arr[j+1] = arr[j]
    return find_insert_pos(arr,temp,j-1,n)

def recursive_insertion_sort_asc(arr, i, n):
    if i==n:
        return
    j = i-1 
    temp=arr[i]
    # since j: int (immutable) : pass by obj value, so need to return updated j
    j = find_insert_pos(arr,temp,j,n)
    # insert
    arr[j+1] = temp

    recursive_insertion_sort_asc(arr,i+1,n)

arr = [13,46,24,52,20,9]
print(f'Before : {arr}')
recursive_insertion_sort_asc(arr,1,len(arr))
print(f'after : {arr}')
```
