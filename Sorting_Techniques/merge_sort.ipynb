### merge sort ascending 

```python
def merge(arr,left,mid,right):
    # additional arr : in python we can just append also
    # sorted_subarr = [None for _ in range(right-left+1)]
    sorted_subarr = []
    i,j = left,mid+1

    while i<=mid and j<=right:
        if arr[i] <= arr[j]:
            sorted_subarr.append(arr[i])
            i+=1
        else:
            sorted_subarr.append(arr[j])
            j+=1
    
    # left subarr : complete, only right subarr : remains
    if i> mid:
        while j<=right:
            sorted_subarr.append(arr[j])
            j += 1

    # right subarr : complete, only left subarr : remains
    if j>right:
        while i<=mid:
            sorted_subarr.append(arr[i])
            i+= 1

    # Update sorted subarr into original array (since NO return in base case,
    # need to modify arr inplace of returning arr and then updating it)

    arr[left:right+1] = sorted_subarr




def merge_sort_asc(arr, left, right):
    
    mid = (left + right) // 2

    # base case : >1 ele
    if left<right:
        # divide left 1/2
        merge_sort_asc(arr,left,mid)
        # divide right 1/2
        merge_sort_asc(arr,mid+1,right)

        # conquer : merge 2 1/2(s)
        merge(arr,left,mid,right)

arr = [4,2,1,6,7]
print(f'before : {arr}')
merge_sort_asc(arr,0,len(arr)-1)
print(f'after : {arr}')
```
