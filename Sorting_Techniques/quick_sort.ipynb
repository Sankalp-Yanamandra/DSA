### quick sort ascending ; 1st ele as pivot

```python
def partition(arr, low, high):
    # First element as pivot
    pivot = arr[low]
    start, end = low, high

    while True:

        # Find element greater than pivot from the left
        while start <= high and arr[start] <= pivot:
            start += 1

        # Find element less than or equal to pivot from the right
        while end >= low and arr[end] > pivot:
            end -= 1

        # If pointers have not crossed
        if start <= end:
            arr[start], arr[end] = arr[end], arr[start]
        else:
            # Put pivot in its correct position
            arr[end], arr[low] = arr[low], arr[end]
            return end


def quick_sort_asc(arr, low, high):
    if low < high:
        loc = partition(arr, low, high)

        # Sort left side
        quick_sort_asc(arr, low, loc - 1)

        # Sort right side
        quick_sort_asc(arr, loc + 1, high)


arr = [4, 6, 2, 5, 7, 9, 1, 3]

print(f'before : {arr}')

quick_sort_asc(arr, 0, len(arr) - 1)

print(f'after : {arr}')
```
