## Selection sort Ascending

```python
def selectionSortAsc(arr):
    n = len(arr)
    # outer loop : traversing the unsorted sublist; until only last ele in unsorted sublist
    for i in range(0, n-1):
        # inner loop : find minm in unsorted sublist N swap with 1st ele of unsorted sublist
        
        # let 1st ele of unsorted sublist : minm
        min_index = i
        # find minm among all
        for j in range(i+1,n):
            if arr[min_index] > arr[j]:
                # update minm
                min_index = j
        
        if min_index != i:
            #swap
            arr[min_index], arr[i] = arr[i], arr[min_index]

arr = list(map(int,input().split()))
print(f'before : {arr}')
selectionSortAsc(arr)
print(f'after : {arr}')
```
