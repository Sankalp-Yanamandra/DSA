```python
'''
Problem Statement: Given an array,
 we have found the number of occurrences of each element in the array.
'''

def countOccurrences(arr):
    frequency = {}

    for element in arr:
        if element in frequency:
            frequency[element] += 1
        else:
            frequency[element] = 1

    for key,value in frequency.items():
        print(f'{key} : {value}')

arr = list(map(int, input().split()))
countOccurrences(arr)


```

```python
'''
Find the highest/lowest frequency element
Problem Statement: Problem Statement: Given an array of size N.
Find the highest and lowest frequency element.
'''

def getMostLeastFrequent(arr):
    frequency = {}

    for element in arr:
        if element in frequency:
            frequency[element] += 1
        else:
            frequency[element] = 1


    maxfreq = max(frequency.values())
    minfreq = min(frequency.values())

    for key,value in frequency.items():
        if value == maxfreq:
            print(f'Max freq ele : {key}')
        elif value == minfreq:
            print(f'Min freq ele : {key}')

arr = list(map(int, input().split()))
getMostLeastFrequent(arr)
        
```
